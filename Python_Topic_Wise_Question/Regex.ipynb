```python
import re
```

```python
# Q1. Given

text = "Aman 20 Rahul 25 Neha 30"

# use re.findall() with \d+ to extract all numbers.

result=re.findall(r"\d+",text)
print(result)
```

```python
# Q2. Given 

text = "cat dog cat rat cat"

# find all occurrences of "cat" and print the list.

result=re.findall(r"cat",text)
print(result)
```

```python
# Q3. Given 

text = "Python is easy and Python is powerful"

# find all occurrences of "Python".

result=re.findall(r"Python",text)
print(result)
```

```python
# Q4. Given

text = "A1 B2 C3 D4"

# extract all digits

result=re.findall(r"\d+",text)
print(result)
```

```python
# Q5. Given

text = "apple123 mango456 BANANA789"

# extract all sequences of lowercase letters

result=re.findall(r"[a-z]",text)
print(result)
```

```python
# Q6. Given

text = "Aman scored 85, Rahul scored 92, Neha scored 78"

# extract all marks as a list

result=re.findall(r"\d+",text)
print(result)
```

```python
# Q7. Given

text = "2024-10-12, 2025-05-20, 2026-10-17"

# extract all dates in the format YYYY-MM-DD.

result=re.findall(r"\d{4}-\d{2}-\d{2}",text)
print(result)
```

```python
# Q8. Given 

text = "Emails: aman@gmail.com, rahul@yahoo.com, neha@outlook.com"

# extract all email addresses using a suitable regex pattern.

result=re.findall(r"\w+@\w+\.com",text)
print(result)
```

```python
# Given 

text = "Order A123, Order B456, Order C789"

# extract all order IDs

result=re.findall(r"[A-Z]\d+",text)
print(result)
```

```python
# Q10. Given

text = "Phone: 98761019102292543210, Alternate: 876543211292929292209, Emergency: 76543210910380480480998"

# extract all 10-digit phone numbers

result=re.findall(r"\d{10}",text)
print(result)
```
