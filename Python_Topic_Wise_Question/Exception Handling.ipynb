```python
# Q1. Write a program that takes two numbers from the user and divides the first number by the second.

try:
    num1=int(input("Enter the Number 1 : "))
    num2=int(input("Enter the Number 2 : "))

    result=num1/num2
except ZeroDivisionError:
    print("Can't Divisible By Zero : ")
else:
    print(result)
```

```python
# Q4. Write a program that asks the user for two numbers and performs division.
# Use:
# - try for the calculation
# - except for errors
# - else to print "Division successful" when no exception occurs.

try:
    num1=int(input("Enter the Number 1 : "))
    num2=int(input("Enter the Number 2 : "))

    result = num1/num2

except ZeroDivisionError:
    print("Not Divisible By Zero : ")

except ValueError :
    print("Enter The Valid Number : ")
else:
    print("Result = ",round(result,2))
```

```python
# Q5. Write a program that asks the user to enter a number.
# Use try-except-finally so that:
# - If the input is valid, print the number.
# - If invalid, print "Invalid input".
# - finally should always print "Program completed".

try:
    num=int(input("Enter The Number : "))
    print(num)

except ValueError : 
    print("Inalid Number : ")

finally:
    print("Program Completed : ")
```
