```python
# Create a program that stores student records in a file named students.txt.
# Each record should contain a student's name and marks, for example:
# Aman,85
# Rahul,72
# Riya,91


# Your program should:
# 1. Accept a student's name and marks from the user.
# 2. Append the new record to the file.
# 3. Read all records from the file.
# 4. Calculate the average marks of all students.
# 5. Display the student with the highest marks.



# Step 1: User se student ki details lo

name = input("Enter student name: ")
marks = int(input("Enter student marks: "))

# Step 2: File mein new record append karo

with open("students.txt", "a") as file:
    file.write(f"{name},{marks}\n")

# Step 3: Saare records read karo
with open("students.txt", "r") as file:
    records = file.readlines()

# Step 4: Records ko process karo

students = []

for record in records:
    try:
        name, marks = record.strip().split(",")
        marks = int(marks)
        students.append((name, marks))
    except ValueError:
        continue

# Step 5: Average marks calculate karo

total = 0

for student in students:
    total += student[1]

average = total / len(students)

# Step 6: Highest marks wala student find karo

topper = max(students, key=lambda student: student[1])

# Step 7: Results print karo

print("\nAll Students:", students)
print("Average Marks:", average)
print("Highest Marks:", topper[0])
print("Highest Marks Obtained:", topper[1])
```
