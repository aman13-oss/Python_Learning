```python
# Q6. Take two numbers from the user and print the results of all arithmetic operations:

# +
# -
# *
# /
# %
# //
# **

a=int(input("Enter the Number1: "))
b=int(input("Enter the Number2: "))

sum=a+b
sub=a-b
prod=a*b
div=a/b
rem=a%b
floor_dividion=a//b
power_of_two=a**b

print("sum= ",sum,"\n","sub= ",sub,"\n","prod= ",prod,"\n","div= ",div,"\n","rem= ",rem,"\n","floor_division= ",floor_dividion,"power_of_two= ",power_of_two)
```

```python
# Q7. A student has marks in three subjects. Store the marks in three variables and calculate:

# Total marks
# Average marks
# Percentage

# Assume each subject is out of 100.

marks1=int(input("Enter the marks1: "))
marks2=int(input("Enter the marks2: "))
marks3=int(input("Enter the marks3: "))

Total_Marks=marks1+marks2+marks3
Average_Marks=round((marks1+marks2+marks3)/3,)
Percentage=round((Total_Marks*100)/300,2)

print("Total_Marks= ",Total_Marks,"\n","Average_Marks= ",Average_Marks,"\n", "Percentage",Percentage)
```
