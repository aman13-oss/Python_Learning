```python
# Q7. A student has marks in three subjects. Store the marks in three variables and calculate:

# Total marks
# Average marks
# Percentage

# Assume each subject is out of 100.

marks1=int(input("Enter the marks1: "))
marks2=int(input("Enter the marks2: "))
marks3=int(input("Enter the marks3: "))

Total_Marks=marks1+marks2+marks3
Average_Marks=round((marks1+marks2+marks3)/3,)
Percentage=round((Total_Marks*100)/300,2)

print("Total_Marks= ",Total_Marks,"\n","Average_Marks= ",Average_Marks,"\n", "Percentage",Percentage)
```

```python
# Q9. Take the principal, rate, and time from the user and calculate:

# Simple Interest = (P × R × T) / 100
# Then calculate the total amount = Principal + Simple Interest.
# Print all values clearly.

P = int(input("Enter the principal Amount: "))
R = int(input("Enter the Rate: "))
T = int(input("Enter the Time: "))

SI = (P * R * T) / 100
total_Amount = P + SI

print("Simple Interest =", SI)
print("Total Amount =", total_Amount)
```

```python
# Q10. Take a 3-digit number from the user and use arithmetic operators to find:

# The hundreds digit
# The tens digit
# The ones digit
# Sum of all three digits
# Product of all three digits

# Condition: Do not convert the number into a string.

num=int(input("Enter the number"))
hundred=num//100
tens=(num//10)%10
ones=num%10

sum=hundred+tens+ones
product=hundred*tens*ones


print("Hundred digit= ",hundred)
print("Tens digit= ",tens)
print("Ones= ",ones)
print("Sum of all three digits= ",sum)
print("Product of all three digts= ",product)
```
