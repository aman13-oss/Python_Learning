# CASE STUDY 01 — Digital Queue Token Manager

```python
# A service centre issues tokens for three counters. A daily token log may contain repeated token numbers because a
# scanner occasionally registers the same token twice. The manager needs the first occurrence of each token and a
# compact report.
# Given data

# tokens = [31, 25, 31, 42, 25, 50, 42, 61]

# Requirements
# 1. Display the unique token numbers in the order they first appeared.
# 2. Identify how many duplicate scans occurred.
# 3. Display the smallest and largest valid token numbers.
# Hint
# Which structure remembers previously encountered tokens? What should be kept in a list to preserve arrival order?

tokens = [31, 25, 31, 42, 25, 50, 42, 61]

seen=set()
unique=[]
duplicate=0

for token in tokens:
    if token in seen:
        duplicate +=1
    else:
        seen.add(token)
        unique.append(token)

print("Unique Tokes : ",unique)
print("Duplicate Tokens : ",duplicate)
print("Largest token : ",max(unique))
print("Smallest Token : ",min(unique))
```

# CASE STUDY 02 — Museum Entry Slot Checker

```python
# A museum accepts visitors in morning and afternoon slots. A booking record contains the chosen slot and the number of
# tickets. Visitors requesting more than four tickets in one booking must be declined; the centre also wants to know which
# slot received more approved visitors.
# Given data
# bookings = [("morning", 2), ("afternoon", 5), ("morning", 3), ("afternoon", 4)]

# Requirements

# 1. Accept only bookings with quantities from 1 to 4.
# 2. Display Accepted or Rejected for each booking.
# 3. Show the total approved tickets for morning and afternoon, and name the busier slot (or Tie).
# Hint
# Can you maintain separate counters for the two slots? When should a rejected booking be excluded?

bookings = [("morning", 2), ("afternoon", 5), ("morning", 3), ("afternoon", 4)]

morning=0
afternoon=0
for slot,tickets in bookings:
    if tickets>=1 and tickets<=4:
        print("Accepted: ")

        if (slot=="morning"):
            morning +=tickets
        else:
            afternoon+= tickets
    else:
        print("Rejected: ")

print("Morning",morning)
print("afternoon",afternoon)

if(morning > afternoon):
    print("Morning Slot")
elif (afternoon > morning):
    print("Afternoon Slot ")
else:
    print("Tie")
```

# CASE STUDY 03 — Parcel Label Formatter

```python
# A warehouse receives parcel labels written with inconsistent spacing and capitalization. For an internal display, each
# label must be normalized and converted to a short code made from the first three letters of each word.
# Given data
# labels = [" blue box ", "RED crate", " green bag"]

# Requirements

# 1. Remove leading/trailing whitespace from each label.
# 2. Convert the cleaned label to title case.
# 3. Build and display a hyphen-separated code using up to the first three uppercase letters of each word (for example,
# Blue Box becomes BLU-BOX).
# Hint
# Which string methods remove extra whitespace and separate words? Can slicing take the first three letters?

labels = [" blue box ", "RED crate", " green bag"]

for label in labels:
    label=label.strip().title()
    word=label.split()
    code=word[0][:3].upper() + "-"+ word[1][:3].upper()
    print(label,"->",code)
```

# CASE STUDY 04 — Water Tank Reading Summary

```python
# A building sensor sends hourly water-level readings as text. Some readings are corrupted. Valid readings are whole
# numbers from 0 to 100 inclusive. The maintenance team needs a report without allowing corrupted values to crash the
# program.
# Given data
# readings = ["55", "bad", "101", "70", "-4", "85"]
# Requirements
# 1. Identify and print valid readings.
# 2. Count rejected entries, including non-numbers and out-of-range values.
# 3. Display the average of the valid readings rounded to one decimal place. If none are valid, display No valid readings.
# Hint
# Convert each reading carefully; which exception might occur? Only calculate the average if the valid list is non-empty.

readings = ["55", "bad", "101", "70", "-4", "85"]
valid=[]
rejected=0

for reading in readings:
    try:
        num=int(reading)

        if(num >=0 and num <=100):
            valid.append(num)
            print("Valid",num)
        else:
            rejected +=1

    except ValueError:
        rejected +=1

print("Rejected Enteries: ",rejected)

if len(valid)>0:
    avg=sum(valid)/len(valid)
    print("Average: ",round(avg,1))
else:
    print("No Valid Reading: ")
```

# CASE STUDY 05 — School Timetable Clash Detector

```python
# A college timetable lists practical sessions as day-and-period pairs. Two sessions assigned to the same pair create a
# clash. The coordinator wants to report the occupied slots and detect repeated assignments.
# Given data
# sessions = [("Mon", 2), ("Tue", 1), ("Mon", 2), ("Wed", 3), ("Tue", 1)]
# Requirements
# 1. Display each distinct day-period slot in the order first encountered.
# 2. Display each clashing slot once.
# 3. Show the number of distinct occupied slots and distinct clashing slots.
# Hint
# A pair can be stored as a tuple. Which data structure allows fast checks for repeated pairs?

sessions = [("Mon", 2), ("Tue", 1), ("Mon", 2), ("Wed", 3), ("Tue", 1)]

occupied=[]
clashes=set()

for session in sessions:
    if session in occupied:
        clashes.add(session)
    else:
        occupied.append(session)


print("Occupied slots:", occupied)
print("Clashing slots:", clashes)
print("Distinct occupied slots:", len(occupied))
print("Distinct clashing slots:", len(clashes))
```
